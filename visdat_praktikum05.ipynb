## **Scatter Plot dengan Matplotlib**

```python
import matplotlib.pyplot as plt

# Data dummy
suhu = [20, 22, 24, 26, 28, 30, 32, 34, 36]
penjualan = [50, 60, 70, 90, 100, 110, 130, 150, 180]

# Membuat scatter plot
plt.scatter(suhu, penjualan, color='pink')
plt.title('Hubungan Penjualan Es Krim dan Suhu')
plt.xlabel('Suhu')
plt.ylabel('Penjualan Es Krim')
plt.show()
```

## **Kustomisasi Scatter Plot**

```python
import matplotlib.pyplot as plt

# Data dummy
suhu = [20, 22, 24, 26, 28, 30, 32, 34, 36]
penjualan = [50, 60, 70, 90, 100, 110, 130, 150, 180]

# Membuat scatter plot
plt.scatter(suhu, penjualan, color='orange', s=100, edgecolors='black', alpha=0.7)
plt.title('Hubungan Penjualan Es Krim dan Suhu')
plt.xlabel('Suhu')
plt.ylabel('Penjualan Es Krim')
plt.grid(True)
plt.show()
```

## **Menambahkan Scatter Plot**

```python
import matplotlib.pyplot as plt

# Data dummy
suhu = [20, 22, 24, 26, 28, 30, 32, 34, 36]

# Data tambahan untuk kategori hari
# Menambahkan satu elemen ke penjualan_kerja agar panjangnya sama dengan suhu dan penjualan_akhir_pekan
penjualan_kerja = [50, 60, 70, 80, 90, 100, 110, 130, 120] # Contoh penambahan data
penjualan_akhir_pekan = [60, 70, 80, 100, 110, 120, 140, 160, 200]

#Scatter plot multiple
plt.scatter(suhu, penjualan_kerja, color='pink', label='Hari Kerja', s=80)
plt.scatter(suhu, penjualan_akhir_pekan, color='purple', label='Akhir Pekan', s=80)
plt.title('Penjualan Es Krim Berdasarkan Hari')
plt.xlabel('Suhu')
plt.ylabel('Penjualan Es Krim')
plt.legend()
plt.show()
```

## **Scatter Plot untuk Melihat Data dengan Beberapa Variabel**

```python
import matplotlib.pyplot as plt

# Data dummy
suhu = [20, 22, 24, 26, 28, 30, 32, 34, 36]
penjualan = [50, 60, 70, 90, 100, 110, 130, 150, 180]

# Data tambahan
kelembapan = [60, 65, 70, 75, 80, 85, 90, 95, 100]
ukuran_titik = [200, 220, 240, 260, 280, 300, 320, 340, 360]

#Scatter plot dengan ukuran dan warna yang bervariasi
plt.scatter(suhu, penjualan, c=kelembapan, s=ukuran_titik, cmap='viridis', alpha=0.6)
plt.colorbar(label='Kelembapan')
plt.title('Hubungan Penjualan, Suhu, dan Kelembapan')
plt.xlabel('Suhu')
plt.ylabel('Penjualan Es Krim')
plt.colorbar(label='Kelembapan')
plt.show()
```
