### **PIE CHART**

### **a. Pie Chart Sederhana**

```python
import matplotlib.pyplot as plt

# Data partisipasi dosen
kegiatan = ['Olahraga', 'Seni', 'Musik', 'Debat']
persentase = [30, 20, 25, 25]

# Membuat pie chart
plt.pie(persentase, labels=kegiatan, autopct='%1.1f%%', startangle=140)
plt.title('Persentase Partisipasi Siswa dalam Kegiatan Ekstrakurikuler')
plt.show()
```

### **b. Pie Chart dengan Matplotlib**

```python
import matplotlib.pyplot as plt

# Data partisipasi siswa
kegiatan = ['Olahraga', 'Seni', 'Musik', 'Debat']
persentase = [30, 20, 25, 25]

# Warna khusus
colors = ['gold', 'lightblue', 'lightcoral', 'limegreen']

# Membuat pie chart dengan warna
plt.pie(persentase, labels=kegiatan, colors=colors, autopct='%1.1f%%', startangle=140)
plt.title('Persentase Partisipasi Siswa dengan Warna Kustom')
plt.show()
```

### **c. Kustomisasi Pie Chart**

```python
import matplotlib.pyplot as plt

# Data partisipasi siswa
kegiatan = ['Olahraga', 'Seni', 'Musik', 'Debat']
persentase = [30, 20, 25, 25]

# Warna khusus
colors = ['gold', 'lightblue', 'lightcoral', 'limegreen']

explode = [0.1, 0, 0, 0]  # Bagian pertama (Olahraga) sedikit dikeluarkan

plt.pie(persentase, labels=kegiatan, colors=colors, autopct='%1.1f%%', startangle=140,
        explode=explode, shadow=True)
plt.title('Persentase Partisipasi Siswa')
plt.legend(kegiatan, loc='best')
plt.show()
```

### **d. Multiple Horizontal Bar Chart**

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# Data kelompok 1
kegiatan_1 = ['Olahraga', 'Seni']
persentase_1 = [40, 60]

# Data kelompok 2
kegiatan_2 = ['Musik', 'Debat']
persentase_2 = [50, 50]

# Pie chart untuk kelompok 1
axs[0].pie(persentase_1, labels=kegiatan_1, autopct='%1.1f%%', startangle=140)
axs[0].set_title('Partisipasi Kelompok 1')

# Pie chart untuk kelompok 2
axs[1].pie(persentase_2, labels=kegiatan_2, autopct='%1.1f%%', startangle=140)
axs[1].set_title('Partisipasi Kelompok 2')

plt.suptitle('Perbandingan Partisipasi Siswa')
plt.show()
```

### **AREA CHART**

### **a. Area Chart dengan Matplotlib**

```python
# Modul Visualisasi Penjualan Bulanan - Area Chart
import matplotlib.pyplot as plt
import numpy as np

# Data Penjualan Bulanan
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
shoes = [500, 600, 700, 800, 650, 700, 800, 850, 900, 750, 800, 1000]

# Membuat Area Chart Dasar
plt.figure(figsize=(10, 6))
plt.fill_between(months, shoes, color='blue', alpha=0.4, label="Shoes")
plt.title("Basic Area Chart: Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Units Sold")
plt.legend()
plt.show()
```

### **b. Kustomisasi Area Chart**

```python
# Modul Visualisasi Penjualan Bulanan - Area Chart
import matplotlib.pyplot as plt
import numpy as np

# Data Penjualan Bulanan
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
shoes = [500, 600, 700, 800, 650, 700, 800, 850, 900, 750, 800, 1000]
sandals = [300, 350, 400, 450, 500, 300, 600, 400, 700, 750, 500, 850]

plt.figure(figsize=(10, 6))
plt.fill_between(months, shoes, color='blue', alpha=0.5, label="Shoes")
plt.fill_between(months, sandals, color='green', alpha=0.5, label="Sandals")
plt.title("Customized Area Chart: Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)  # Memutar label bulan
plt.grid(axis='y', linestyle='--', alpha=0.6)  # Garis grid horizontal
plt.legend()
plt.show()
```

### **c. Multiple Area Chart**

```python
# Modul Visualisasi Penjualan Bulanan - Area Chart
import matplotlib.pyplot as plt
import numpy as np

# Data Penjualan Bulanan
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
shoes = [500, 600, 700, 800, 650, 700, 850, 900, 750, 800, 950, 1000]
sandals = [300, 350, 400, 450, 500, 500, 600, 650, 700, 750, 800, 850]
socks = [200, 250, 300, 350, 400, 400, 450, 500, 600, 700, 750, 800]

# Convert lists to numpy arrays
shoes_np = np.array(shoes)
sandals_np = np.array(sandals)
socks_np = np.array(socks)

# Multiple Area Chart
plt.figure(figsize=(12, 6))
plt.fill_between(months, shoes_np, color="blue", alpha=0.4, label="Shoes")
plt.fill_between(months, np.add(shoes_np, sandals_np), shoes_np, color="green", alpha=0.4, label="Sandals")
plt.fill_between(months, np.add(np.add(shoes_np, sandals_np), socks_np), np.add(shoes_np, sandals_np), color="orange", alpha=0.4, label="Socks")

plt.title("Multiple Area Chart: Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Units Sold")
plt.legend()
plt.show()
```
